```python
import pandas as pd

# Real Indian state population data based on Census 2011
data = {
    "State": [
        "Uttar Pradesh", "Maharashtra", "Bihar", "West Bengal",
        "Madhya Pradesh", "Tamil Nadu", "Rajasthan", "Karnataka",
        "Gujarat", "Andhra Pradesh", "Odisha", "Telangana",
        "Kerala", "Jharkhand", "Assam"
    ],
    "Population": [
        199812341, 112374333, 104099452, 91276115,
        72626809, 72147030, 68548437, 61095297,
        60439692, 49577103, 41974218, 35003674,
        33406061, 32988134, 31205576
    ],
    "Area_km2": [
        240928, 307713, 94163, 88752, 308252,
        130060, 342239, 191791, 196024, 162968,
        155707, 112077, 38863, 79716, 78438
    ]
}

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# Filter: Select states with population greater than 50 million
high_population = df[df["Population"] > 50000000]

print("States with population greater than 50 million:")
display(high_population)
```

```python
# GroupBy: Calculate the average population by area category
df["Area_Category"] = pd.cut(
    df["Area_km2"],
    bins=[0, 100000, 200000, 400000],
    labels=["Small", "Medium", "Large"]
)

grouped = df.groupby("Area_Category", observed=True)["Population"].mean()

print("Average population by area category:")
display(grouped)
```

```python
# Create a second DataFrame with region information
region_data = pd.DataFrame({
    "State": [
        "Uttar Pradesh", "Maharashtra", "Bihar", "West Bengal",
        "Madhya Pradesh", "Tamil Nadu", "Rajasthan", "Karnataka",
        "Gujarat", "Andhra Pradesh", "Odisha", "Telangana",
        "Kerala", "Jharkhand", "Assam"
    ],
    "Region": [
        "North", "West", "East", "East",
        "Central", "South", "North", "South",
        "West", "South", "East", "South",
        "South", "East", "North-East"
    ]
})

# Merge: Combine the population DataFrame with region information using State
merged_df = pd.merge(df, region_data, on="State", how="left")

print("Merged DataFrame:")
display(merged_df.head(10))
```

```python
# Pivot Table: Summarize total population by region and area category
pivot = pd.pivot_table(
    merged_df,
    values="Population",
    index="Region",
    columns="Area_Category",
    aggfunc="sum",
    fill_value=0
)

print("Population by region and area category:")
display(pivot)
```

```python
# Export: Save the cleaned DataFrame as CSV and Parquet files

# Save as CSV
merged_df.to_csv("cleaned_indian_population.csv", index=False)

# Save as Parquet
merged_df.to_parquet("cleaned_indian_population.parquet", index=False)

print("CSV and Parquet files created successfully!")
```

```python
# Compare: Check the file sizes of CSV and Parquet

import os

csv_size = os.path.getsize("cleaned_indian_population.csv")
parquet_size = os.path.getsize("cleaned_indian_population.parquet")

print(f"CSV file size: {csv_size} bytes")
print(f"Parquet file size: {parquet_size} bytes")

if csv_size > parquet_size:
    print("Parquet file is smaller than CSV.")
else:
    print("CSV file is smaller than Parquet.")
```
