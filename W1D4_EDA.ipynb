```python
import pandas as pd

data = {
    "Name": ["Asha", "Rahul", "Priya", "Arun", "Sneha", "Vikram", "Anjali", "Kiran", "Meena", "Ravi"],
    "Age": [21, 23, 22, 25, 21, 24, 23, 26, 22, 24],
    "City": ["Bangalore", "Chennai", "Bangalore", "Mumbai", "Chennai",
             "Bangalore", "Mumbai", "Chennai", "Bangalore", "Mumbai"],
    "Study_Hours": [4, 6, 5, 8, 3, 7, 6, 9, 4, 5],
    "Attendance": [85, 90, 88, 95, 78, 92, 89, 96, 82, 87],
    "Score": [72, 85, 78, 91, 65, 88, 84, 94, 70, 76]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

df.hist(figsize=(12, 8), bins=5)

plt.suptitle("Distribution of Numeric Columns")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

correlation = df.select_dtypes(include="number").corr()

plt.figure(figsize=(8, 6))
plt.imshow(correlation, cmap="coolwarm", interpolation="none")
plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(j, i, f"{correlation.iloc[i, j]:.2f}",
                 ha="center", va="center")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

city_counts = df["City"].value_counts().head(10)

city_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Top-10 City Counts")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## EDA Narrative

The dataset contains information about 10 students, including their age, city, study hours, attendance, and academic score. The numerical variables are Age, Study_Hours, Attendance, and Score, while Name and City are categorical variables. The dataset contains no missing values, as confirmed by the null-value check. The distribution plots show that the students' ages range from 21 to 26 years, while study hours range from 3 to 9 hours. Attendance and scores also show variation across the students.

The correlation analysis shows strong positive relationships among the numerical variables. Study hours and score have a correlation of 0.98, while attendance and score also have a correlation of 0.98. This suggests that students with higher study hours and attendance tend to have higher scores in this sample. The city distribution shows that Bangalore has the highest number of students, followed by Chennai and Mumbai.

One suspicious aspect is the unusually strong correlations between almost all numerical variables. This may be caused by the small dataset and the way the sample data was created. Since the dataset contains only 10 records, the findings should not be generalized to a larger population. For a real ML project, a larger dataset should be collected, checked for outliers and duplicates, and validated before modeling.
