```python
# W1D3: Data Loading, Cleaning & Inspection

import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Load the Indian population dataset
url = "https://raw.githubusercontent.com/Manasa23BEIS123/W1D1-Numpy/main/W1D2/cleaned_indian_population.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# Check for missing values
print("Missing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# Clean column names for easier use
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
# Remove duplicate rows
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
# Fill missing numeric values with the column median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Inspect the cleaned dataset
print("Cleaned dataset shape:", df.shape)

print("\nCleaned dataset:")
display(df.head(10))

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Save the cleaned dataset as CSV
output_file = "W1D3_cleaned_dataset.csv"

df.to_csv(output_file, index=False)

print(f"Cleaned dataset saved successfully as: {output_file}")
```

```python
# Validate the cleaned dataset

# Test 1: Dataset should not be empty
assert len(df) > 0, "Dataset is empty"

# Test 2: No duplicate rows should remain
assert df.duplicated().sum() == 0, "Duplicate rows found"

# Test 3: No missing values should remain in numeric columns
numeric_columns = df.select_dtypes(include=np.number).columns
assert df[numeric_columns].isnull().sum().sum() == 0, "Missing numeric values found"

print("All validation tests passed successfully!")
```
