```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 6, 4, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 95, 88, 72, 98],
    "Marks": [45, 50, 58, 65, 70, 78, 85, 75, 55, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Marks"], bins=5, kde=True)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

correlation = df[["Study_Hours", "Attendance", "Marks"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.lineplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    marker="o"
)

plt.title("Study Hours vs Marks Trend")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Marks"
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.show()
```

## Conclusion

The visualisations show the relationship between study hours, attendance, and student marks. The scatter plots indicate that marks generally increase as study hours and attendance increase. The correlation heatmap also helps identify the strength of relationships between numerical variables. These visualisations demonstrate how Matplotlib and Seaborn can be used during Exploratory Data Analysis (EDA) to identify patterns and trends before applying machine learning models.
