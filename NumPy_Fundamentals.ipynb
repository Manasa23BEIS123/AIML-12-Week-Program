```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
```

```python
# 1D Array
array_1d = np.array([1, 2, 3, 4, 5])

# 2D Array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\n2D Array:")
print(array_2d)
print("Shape:", array_2d.shape)

print("\n3D Array:")
print(array_3d)
print("Shape:", array_3d.shape)
```

```python
# Broadcasting example

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

values = np.array([1, 2, 3])

result = arr + values

print("Original array:")
print(arr)

print("\nValues added using broadcasting:")
print(values)

print("\nResult:")
print(result)
```

```python
numbers = np.array([1, 2, 3, 4, 5])

squared = numbers ** 2
doubled = numbers * 2
added = numbers + 10

print("Original:", numbers)
print("Squared:", squared)
print("Doubled:", doubled)
print("Added 10:", added)
```

```python
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

matrix_result = np.matmul(matrix_a, matrix_b)

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nMatrix multiplication result:")
print(matrix_result)
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
numeric_data = df[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
]

mean_values = numeric_data.mean()
std_values = numeric_data.std()

print("Mean:")
print(mean_values)

print("\nStandard Deviation:")
print(std_values)
```

```python
correlation_matrix = np.corrcoef(
    numeric_data.to_numpy().T
)

print("Correlation Matrix:")
print(correlation_matrix)
```
